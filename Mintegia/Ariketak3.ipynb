### 1
`contains_any(w,s)` funtzio bat idatz ezazu, `True` bueltatuko duena `w` karaktere katearen barnean `s` karaktere kateko letrarik balego eta `False` kontrako kasuan.

```python
def contains_any(w,s):
    b = 0
    for k in s :
        b = b + w.count(k)
    if b > 0 :
        return True
    else :
        return False
def contains_any(w,s):
    for k in s :
        if w.count(k)>0 :
            return True
    return False
def contains_any(w,s):
    for k in s :
        if k in w :
            return True
    return False
```

### 2
`contains_only(w,s)` funtzio bat idatz ezazu, `True` bueltatuko duena `w` karaktere katearen barnean soilik `s` karaktere kateko letrak agertzen badira eta `False` kontrako kasuan.

```python
def contains_only(w,s):
    for k in s :
        if k in w :
            return True
    return False
```
