```python
def fak(n):
    b = 1
    for i in range(2,n+1):
        b = b * i
    return b
    
def exp(x, n=10):
    b = 1
    for i in range(1,n+1):
        s = x**i / fak(i)
        b = b + s
    return b


print(exp(1))
print(exp(1,100))
```
