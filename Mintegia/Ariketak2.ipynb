```python
def fak(n):
    b = 1
    for i in range(2,n+1):
        b = b * i
    return b
    
def exp(x, n=10):
    b = 1
    for i in range(1,n+1):
        s = x**i / fak(i)
        b = b + s
    return b


print(exp(1))
print(exp(1,100))
```

```python
def exp(x, n=10):
    b = 1
    s = 1
    for i in range(1,n+1):
        s = s * x / i
        b = b + s
    return b

print(exp(1))
print(exp(1,100))
```

```python
def sin(x, n=10):
    s = x
    b = x
    for i in range(3,n*2,2):
        s = -s * (x * x) / ((i-1) * i) 
        b = b + s
    return b

print(sin(0))
print(sin(3.141592/2))
print(sin(3.141592))
print(sin(3.141592*2,10000))
print(sin(90,1000000))
```
