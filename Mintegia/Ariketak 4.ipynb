Funtzio bat sortu, testu fitxategi baten hutsune guztiak ezabatu eta emaitza fitxategi berri batean gordetzen duena.

```python
def f6(bideizena1, bideizena2, kodifikazioa1="utf8", kodifikazioa2="utf8"):
    f1 = open(bideizena1, "r", encoding=kodifikazioa1)
    f2 = open(bideizena2, "w", encoding=kodifikazioa2)
    s = f1.readline()
    while s != "":
        s = s.replace(" ","")
        f2.write(s)
        s = f1.readline()
    f1.close()
    f2.close()
```

```python
def f6(bideizena1, bideizena2, kodifikazioa1="utf8", kodifikazioa2="utf8"):
    f1 = open(bideizena1, "r", encoding=kodifikazioa1)
    f2 = open(bideizena2, "w", encoding=kodifikazioa2)
    for s in f1:
        s = s.replace(" ","")
        f2.write(s)
        #print(s, end="", file=f2)
    f1.close()
    f2.close()
```
