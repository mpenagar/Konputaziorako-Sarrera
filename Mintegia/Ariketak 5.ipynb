```python
def ordenatu1(z1,z2):
    # zerrenden kopiak egin originalak ez aldatzeko
    z1 = z1[:]
    z2 = z2[:]
    z = []
    while z1 and z2 :
        if z1[0] < z2[0] :
            z.append(z1.pop(0))
        else :
            z.append(z2.pop(0))
    z.extend(z1)
    z.extend(z2)
    return z

# Hau ikasgai honetako edukia ez bada ere,
#   z.pop(0) egiteak koste handia du
#   z.pop(-1) edo z.pop() egiteak koste txikia du
# beraz z.pop(0) ebitatuko dugu kasu honetan
def ordenatu2(z1,z2):
    z1 = z1[:]
    z2 = z2[:]
    z1.reverse()
    z2.reverse()    
    z = []
    while z1 and z2 :
        if z1[-1] < z2[-1] :
            z.append(z1.pop())
        else :
            z.append(z2.pop())
    z1.reverse()
    z2.reverse()                
    z.extend(z1)
    z.extend(z2)
    return z
    
def ordenatu3(z1,z2):
    # zerrenden kopiarik ez da behar
    i1 = 0
    i2 = 0
    z = []
    while i1 < len(z1) and i2 < len(z2):
        if z1[i1] < z2[i2] :
            z.append(z1[i1])
            i1 += 1
        else :
            z.append(z2[i2])
            i2 += 1
    z.extend(z1[i1:])
    z.extend(z2[i2:])
    return z

    
z1 = [1,3,5,8,10,23,45,65]
z2 = [0,4,6,8,12,34,45,67,78,89]
print(ordenatu1(z1,z2))
print(ordenatu2(z1,z2))
print(ordenatu3(z1,z2))
z1,z2 = z2,z1
print(ordenatu1(z1,z2))
print(ordenatu2(z1,z2))
print(ordenatu3(z1,z2))
print(z1)
print(z2)

```
