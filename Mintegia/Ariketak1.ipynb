```python
def matrikula(nota,nota1,nota2,nota3):
    if nota >= nota1 :
        print("Maiatzean matrikulatu zaitezke")
    elif nota >= nota2 :
        print("Uztailan matrikulatu zaitezke")
    elif nota >= nota3 :
        print("Irailan matrikulatu zaitezke")
    else :
        print("Ezin zaitezke matrikulatu")    
mezua = """
Graduak ondokoak dira:

  1 - Arte ederrak
  2 - Informatika
  3 - Medikuntza

Zein gradutan egin nahi duzu matrikula [1,2,3]: """
gradua = input(mezua)
while gradua != '1' and gradua != '2' and gradua != '3' :
    gradua = input(mezua)
    
nota = float(input("Zein da zure nota: "))

if gradua == '1' : # Arte Ederrak
    matrikula(nota,6.5,6.0,5.0)
elif gradua == '2' : # Informatika
    matrikula(nota,7.0,6.0,5.0)
else : # Medikuntza
    matrikula(nota,8.5,8.0,7.5)
```

$$n! = 1 \cdot 2 \cdot 3 \cdot \dots \cdot (n-1) \cdot n$$

$$10! = \prod_{i=1}^{n}n$$

```python
def lehena(n):
    if n < 2 :
        return False
    for i in range(2,n):
        if n%i == 0 :
            return False
    return True

# ea ongi dabilen....
for i in range(100):
    if lehena(i):
        print(i, end=" ")
```

```python
def lehena(n):
    if n < 2 :
        return False
    i = 2
    while i*i <= n :
        if n%i == 0 :
            return False
        i += 1
    return True

# ea ongi dabilen....
for i in range(100):
    if lehena(i):
        print(i, end=" ")
```
