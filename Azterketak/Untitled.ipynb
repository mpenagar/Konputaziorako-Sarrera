### 1. Ariketa

```python
def hurrengo_mersenne(n):
    i = 4
    while i <= n :
        i = i * 2
    return i-1
```

Funtzioaren exekuzio adibide bat (hau ez da eskatzen):

```python
for i in range(17) :
    print(i, '-->', hurrengo_mersenne(i))
```

### 2. Ariketa

Programa bat eskatzen denez, ez da beharrezkoa funtziorik sortzea, hala ere ebazpen honetan `programa_nagusia` funtzioa sortuko da:

```python
def programa_nagusia():
    for _ in range(5):
        s = input("Sartu zenbaki bat: ")
        if s == "" : 
            break
        n = int(s)
        e = hurrengo_mersenne(n)
        if e == n :
            print(n, "Mersenne-en zenbaki bat da")
        else :
            print(n, "ez da Mersenne! Hurrengoa", e, "da")
    print("Agur!")
```

Programaren exekuzio adibide bat (hau ez da eskatzen):

```python
programa_nagusia()
```
