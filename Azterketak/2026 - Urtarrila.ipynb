### 1a. Ariketa

```python
def taula_irakurri(bideizena, kodifikazioa="utf8"):
    f = open(bideizena, mode="r", encoding=kodifikazioa)
    z = []
    for ilara in f:
        z.append(list(ilara.rstrip()))
    f.close()
    return z
```

Hau ez da eskatzen, exekuzio adibide bat besterik ez da:

```python
taula = taula_irakurri("taula.txt")
taula
```

### 1b. Ariketa

```python
def ilara_okupazioa(taula, i):
    return taula[i].count("*")

def ilara_beteak(taula):
    indizeak = [0]
    b_max = ilara_okupazioa(taula, 0)
    for i in range(1,len(taula)):
        b = ilara_okupazioa(taula, i)
        if b > b_max:
            indizeak = [i]
        elif b == b_max :
            indizeak.append(i)
    return indizeak
```

```python
ilara_beteak(taula)
```

### 1d Ariketa

```python
def patroiak_bilatu(taula):
    patroiak = []
    for i in range(len(taula)):
        p = "".join(taula[i])
        for z in patroiak :
            if z[0] == p :
                z.append(i)
                break
        else :
            patroiak.append([p,i])
    return patroiak
```

Hau ez da eskatzen, exekuzio adibide bat besterik ez da:

```python
patroiak = patroiak_bilatu(taula)
patroiak
```

### 1e Ariketa

```python
def ohiko_patroia(patroiak):
    ohikoa = patroiak[0][0]
    kop = len(patroiak[0])-1
    for i in range(1,len(patroiak)):
        k = len(patroiak[i])-1
        if k > kop :
            ohikoa = patroiak[i][0]
            kop = k
    return [ohikoa,kop]        
```

```python
ohiko_patroia(patroiak)
```
